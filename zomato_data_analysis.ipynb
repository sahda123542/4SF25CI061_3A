```python
total=df.groupby('votes')['approx_cost(for two people)'].sum().reset_index()
plt.plot(total['votes'],total['approx_cost(for two people)'])
```

```python
df['listed_in(city)'].value_counts().plot(
    kind='bar',
    figsize=(10,4)
)

plt.title('Restaurants by City')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
pd.crosstab(df['online_order'], df['book_table']).plot(kind='bar', stacked=True, figsize=(8,5))
plt.title('Order Status by Payment Method')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='approx_cost(for two people)', y='rate', hue='online_order', palette='viridis')
plt.title('Shipping Fee vs Order Amount')
plt.xlabel('Shipping Fee')
plt.ylabel('Order Amount (rupees)')
plt.show()
```

```python
df['votes'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6,10))
plt.title('vote Share')
plt.ylabel('')
plt.show()
```
